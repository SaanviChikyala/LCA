```python
# List, tuple, dictionary
student_list=["Amit", "Rahul", "Priya", "Sneha"]
print("\nStudent list:")
print(student_list)
print("\nDifferent operations on a list---")
print("\nAppending a list:")
student_list.append("Rashi")
print(student_list)
print("\nInserting elements at specific indexes in a list:")
student_list.insert(2,"Ram")
print(student_list)
print("\nSlicing a list:")
sliced_list=student_list[1:4]
print(sliced_list)
print("\nUsing .index() for knowing the index of any element in a list:")
required_index=student_list.index("Sneha")
print(required_index)
print("\nRemoving elements in a list using .remove():")
student_list.remove("Priya")
print(student_list)
print("\nRemoving a specific element using index in a list using .pop():")
student_list.pop(1)
print(student_list)
print("\nOutputing a specific element using index in a list using .pop():")
popped_element=student_list.pop(2)
print(popped_element)

student_tuple1=("Amit", "Rahul", "Priya", "Sneha","Rahul")
student_tuple2=("Rohit", "Rashi")
print("\nStudent tuple:")
print(student_tuple1)
print(student_tuple2)
print("\nDifferent operations on a tuple---")
print("\nUsing .index() for knowing the index of any element in 1st tuple:")
required_index=student_tuple1.index("Sneha")
print(required_index)
print("\nFinding total no. of times a specific element occurs in 1st tuple:")
print(student_tuple1.count("Rahul"))
print("\nConcatenation of tuples:")
print(student_tuple1+student_tuple2)
print("\nSlicing 1st tuple:")
print(student_tuple1[1:5])
print("\nReversing 1st tuple:")
print(student_tuple1[::-1])



student_dict={
    101:"Amit",
    102:"Rahul",
    103: "Priya",
    104:"Sneha"
}
print("\nStudent dictionary:")
print(student_dict)
print("\nDifferent operations on a dictionary---")
print("\nReplacing an element at a specific key-value:")
student_dict[102]="Rohit"
print(student_dict)
print("\nAdding an element in a dictionary:")
student_dict[105]="Neha"
print(student_dict)
print("\nDeleting a certain element at a specific key-value:")
del student_dict[104]
print(student_dict)
print("\nOutputing a specific element using it's key-value:")
print(student_dict.pop(101))
print("\nRemoving all elements from a dictionary:")
print(student_dict.clear())
```
